```python
lst = [1, 2, 3, 4, 5]

l = len(lst)

srch = 3

while l > 0:
    if lst[l - 1] == srch:
        print(f"Found {srch} at index {l - 1}")
        break
    # print(f"{l = }")
    l -= 1
else:
    print(f"{srch} not found in the list")
```

```python
lst = [1, 2, 3, 4, 5]
ln = len(lst)
srch = 3
for i in range(ln):
    if lst[i] == srch:
        print(f"Found {srch} at index {i}")
        break
else:
    print(f"{srch} not found in the list")
```
